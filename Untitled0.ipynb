```python
import folium
from folium import plugins
import json
import requests

# Initialize the map centered on the world
world_map = folium.Map(location=[20, 0], zoom_start=2)

# Load GeoJSON data of world countries (can be downloaded from various sources or fetched via an API)
url = "https://raw.githubusercontent.com/johan/world.geo.json/master/countries.geo.json"
response = requests.get(url)
geo_data = response.json()

# Add the GeoJSON to the map
folium.GeoJson(
    geo_data,
    name="Countries",
    tooltip=folium.GeoJsonTooltip(fields=["name"], labels=True)
).add_to(world_map)

# Add a layer control and display the map
folium.LayerControl().add_to(world_map)

# Save to an HTML file to view in a browser
world_map.save("world_map.html")
```
